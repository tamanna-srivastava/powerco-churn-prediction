### Data Visualisations

#### 1) Overall Churn Rate

```python
# making the figure
fig, ax = plt.subplots(figsize=(7, 7))         
fig.patch.set_facecolor("#001F3F")               # figure outer background
ax.set_facecolor("#002B5B")                      # plot inner background

# preparing the data
churn_counts = client_data1['churn'].value_counts()   # [0: 13191, 1: 1415]
labels = ['Retained', 'Churned']
colors = ['#F39C12', '#662506']                  # green = retained, red = churned
explode = (0, 0.05)                              

# plotting
wedges, texts, autotexts = ax.pie(
    churn_counts,
    labels=labels,
    colors=colors,
    autopct='%1.1f%%',                           # 1 decimal place float, % sign
    startangle=90,                               # start at 00 hours
    explode=explode,
    shadow=True,                                        # shadow add karo
    wedgeprops={'edgecolor': '#001F3F', 'linewidth': 3}, # border between slices
    textprops={'color': 'white', 'fontsize': 13}
)

# making percentages in bold
for autotext in autotexts:
    autotext.set_fontweight('bold')
    autotext.set_fontsize(13)

# setting up the title
ax.set_title(
    "Overall Customer Churn Rate",
    fontsize=16,
    fontweight='bold',
    color='white',
    pad=12                                       # space between chart and title
)

# fixing the layout
plt.tight_layout()                               
plt.show()
```

Approximately 9.7% of PowerCo's customers churned, compared to 90.3% who were retained. This class imbalance is important to note. It poses the need for the model to be evaluated carefully using metrics beyond simple accuracy, such as precision-recall or AUC-ROC.

#### 2) Churn Rate across Sales Channels

```python
print(client_data1['channel_sales'].value_counts())
print(f"\nTotal distinct channels: {client_data1['channel_sales'].nunique()}")
```

```python
print(client_data1.groupby(['channel_sales', 'churn']).size().unstack(fill_value=0))
```

```python
# filtering relevant columns
channel = client_data1[['id', 'channel_sales', 'churn']]

# removing sales channels with insufficient data
big_channels = client_data1['channel_sales'].value_counts()
big_channels = big_channels[big_channels > 12].index
channel = channel[channel['channel_sales'].isin(big_channels)]

# preparing the data for plotting
channel = (channel.groupby([channel['channel_sales'], channel['churn']])['id']
           .count()
           .unstack(level=1)
           .fillna(0))

channel_churn = (channel.div(channel.sum(axis=1), axis=0) * 100).sort_values(by=[1], ascending=False)

# setting figure
fig, ax = plt.subplots(figsize=(10, 6))
fig.patch.set_facecolor('#001F3F')
ax.set_facecolor('#002B5B')

# plotting 
bars_retained = ax.bar(channel_churn.index, channel_churn[0],
                       color='#FEC44F', label='Retained', edgecolor='black')

bars_churned = ax.bar(channel_churn.index, channel_churn[1],
                      bottom=channel_churn[0],
                      color='#662506', label='Churned', edgecolor='black')

# retained labels
for bar, val in zip(bars_retained, channel_churn[0]):
    ax.text(bar.get_x() + bar.get_width()/2,
            bar.get_height()/2,              # labels in the center for retained
            f'{val:.1f}%',
            ha='center', va='center',
            color='#001F3F', fontweight='bold', fontsize=9)

# churned labels
for bar, val in zip(bars_churned, channel_churn[1]):
    ax.text(bar.get_x() + bar.get_width()/2,
            bar.get_y() + bar.get_height()/2,
            f'{val:.1f}%',
            ha='center', va='center',
            color='white', fontweight='bold', fontsize=9)

# Styling
ax.set_title('Churn Rate by Sales Channel',
             fontsize=16, fontweight='bold', color='white', pad=15)
ax.set_xlabel('Sales Channel', color='white', fontsize=12)
ax.set_xticklabels(channel_churn.index, rotation=30, ha='right', color='white', fontsize=8)
ax.set_ylabel('Percentage (%)', color='white', fontsize=12)
ax.set_ylim(0, 100)
ax.tick_params(colors='white')
ax.spines['bottom'].set_color('white')
ax.spines['left'].set_color('white')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.legend(facecolor='#001F3F', labelcolor='white', fontsize=10)

plt.tight_layout()
plt.show()
```

Sales channel analysis reveals significant variation in churn rates. The highest churn rate for a sales channel is 12% and the lowest is 5.6%. PowerCo should investigate what differentiates high-churn channels in terms of customer demographics, contract terms, or service quality. Channels with insufficient data excluded from visualisation.

#### 3) Tenure wise Churn Rate

```python
# Creating tenure buckets
client_data1['tenure_bucket'] = pd.cut(
    client_data1['num_years_antig'],
    bins=[0, 5, 10, 20],
    labels=['<5 years', '5-10 years', '>10 years']
)

# Churn rate per bucket - observed=True adding it
tenure_churn = (client_data1.groupby('tenure_bucket', observed=True)['churn']
                .mean() * 100).reset_index()
tenure_churn.columns = ['tenure', 'churn_rate']


# Figure
fig, ax = plt.subplots(figsize=(8, 5))
fig.patch.set_facecolor('#001F3F')
ax.set_facecolor('#002B5B')

# Bar plot — hue wali warning fix
sns.barplot(data=tenure_churn, x='tenure', y='churn_rate',
            hue='tenure',
            palette=['#662506', '#D95F0E', '#FEC44F'],
            edgecolor='black', legend=False, ax=ax)

# Value labels - setting loop on all containers
for container in ax.containers:
    ax.bar_label(container, fmt='%.1f%%',
                 color='white', fontweight='bold', fontsize=11, padding=5)

# Styling
ax.set_title('Churn Rate by Customer Tenure', fontsize=15, fontweight='bold', color='white', pad=15)
ax.set_xlabel('Tenure Group', color='white', fontsize=11)
ax.set_ylabel('Churn Rate (%)', color='white', fontsize=11)
ax.tick_params(colors='white')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['bottom'].set_color('white')
ax.spines['left'].set_color('white')

plt.tight_layout()
plt.show()
```

Customers with less than 5 years tenure show the highest churn rate at 11.5%, nearly double compared to the 5-10 year group at 7%. Interestingly, the longest-tenure customers (>10 years) show a slightly higher churn rate of 7.8% than the middle group, suggesting that even long-standing customers are not entirely immune to churn, possibly due to better competitor offers or contract renewals. PowerCo should prioritize retention efforts for new customers in their first 5 years, while also monitoring long-tenure customers at renewal points.

### Distribution Analysis
The following histograms and boxplots examine the distribution of key numerical variables. Most variables exhibit strong positive skewness with extreme outliers, as indicated by the descriptive statistics above. Log transformation will be applied in the Feature Engineering step to address this.

#### 1) Consumption

```python
def plot_hist(dataframe, column, ax):
    temp = pd.DataFrame({
        'Retained': dataframe[dataframe['churn']==0][column],
        'Churned':  dataframe[dataframe['churn']==1][column]
    })
    temp.plot(kind='hist', bins=50, stacked=True, ax=ax,
              color=['#FEC44F', '#662506'])
    ax.set_facecolor('#002B5B')
    ax.set_title(column, color='white', fontsize=11, fontweight='bold')
    ax.tick_params(colors='white')
    ax.ticklabel_format(style='plain', axis='x')
    ax.spines['bottom'].set_color('white')
    ax.spines['left'].set_color('white')
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.legend(facecolor='#001F3F', labelcolor='white', fontsize=9)
```

```python
# CONSUMPTION HISTOGRAMS
fig, axs = plt.subplots(2, 2, figsize=(14, 10))
fig.patch.set_facecolor('#001F3F')
fig.suptitle('Consumption Variables - Distribution by Churn Status',
             fontsize=16, fontweight='bold', color='white', y=1.01)

plot_hist(client_data1, 'cons_12m', axs[0][0])
plot_hist(client_data1[client_data1['has_gas']=='t'], 'cons_gas_12m', axs[0][1])
plot_hist(client_data1, 'cons_last_month', axs[1][0])
plot_hist(client_data1, 'imp_cons', axs[1][1])

plt.tight_layout()
plt.show()
```

All four consumption variables - `cons_12m`, `cons_gas_12m`, `cons_last_month`, and `imp_cons` - exhibit strong positive (right) skewness, with the vast majority of customers clustered at low consumption values and a long tail of extreme outliers. This indicates that while most customers have modest energy usage, a small number of high-consumption customers drive the distribution. Outliers are further examined in the boxplots below. Log transformation will be applied in the Feature Engineering step to address this skewness. Note: churned customers (9.7%) form a small visible stack atop the retained segment, reflecting the class imbalance in the dataset.

```python
def plot_box(dataframe, column, ax):
    sns.boxplot(data=dataframe, x=column, ax=ax,
                color='#FEC44F',           # single warm color
                flierprops=dict(marker='o', markerfacecolor='#662506', 
                                markersize=3, alpha=0.5))  # outlier dots
    ax.set_facecolor('#002B5B')
    ax.set_title(column, color='white', fontsize=11, fontweight='bold')
    ax.tick_params(colors='white')
    ax.ticklabel_format(style='plain', axis='x')
    ax.spines['bottom'].set_color('white')
    ax.spines['left'].set_color('white')
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
```

```python
# CONSUMPTION BOXPLOTS
fig, axs = plt.subplots(2, 2, figsize=(14, 10))
fig.patch.set_facecolor('#001F3F')
fig.suptitle('Consumption Variables - Outlier Analysis',
             fontsize=16, fontweight='bold', color='white', y=1.01)

plot_box(client_data1, 'cons_12m', axs[0][0])
plot_box(client_data1[client_data1['has_gas']=='t'], 'cons_gas_12m', axs[0][1])
plot_box(client_data1, 'cons_last_month', axs[1][0])
plot_box(client_data1, 'imp_cons', axs[1][1])

plt.tight_layout()
plt.show()
```

The boxplots confirm extreme positive skewness across all consumption variables. The interquartile range (IQR) appears as a narrow band near zero, while outliers extend far to the right - indicating a small number of very high-consumption customers. As noted in the descriptive statistics above, the mean of `cons_12m` (159,220), is over 11 times the median (14,115), further confirming the severity of the skew. Similarly, `cons_gas_12m` has a median of zero, meaning more than 50% of customers recorded no gas consumption in the last 12 months. These extreme outliers likely represent large industrial or commercial clients. Log transformation will be applied in Feature Engineering to normalise these distributions.
